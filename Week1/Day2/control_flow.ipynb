# Control Flow
## 1. if/elif/else


```python
grade =85

if(grade >= 55 ):
    print("you passed the exam")
else:
    print("you failed the exam")
    
```

```python
grade = 85
if (grade >= 90):
    print("A")
elif (grade >= 80):
    print("B")
else:
    print("C")  
```

## 2. For loop

```python
# find the odd numbers from 1 to 20
for i in range(1, 21):
    if i % 2 != 0:
        print(i)
    
```

## 3. While

```python
count = 5
sum=0
while count > 0:
    sum += count
    count -= 1
    print("count: " + str(5 - count) + ", value: " + str(count))
    if sum%2==0:
        print("even")
    else:
        print("odd")
print(sum)
```

## 4.  Break & Continue


```python
for i in range(4, 15):
    if i % 6 !=0:
        print(i)
        continue
           
        
    if i % 6 == 0:
        print(i)
        break
    
```
