# List Comprehensions & Object-Oriented Programming (OOP) Basics



In Python, there are shorter and cleaner ways to write code than the usual for loop.

 we will cover two important topics:

1. List Comprehensions — a fast, one-line way to build a new list.
2. Object-Oriented Programming (OOP) Basics — how to group data and behavior together using class.

## 1. List Comprehensions

A list comprehension builds a new list in one readable line, instead of writing a multi-line for loop.

It is considered idiomatic ("Pythonic") code, and you will see it constantly in real data-science work.

`
new_list = [expression for item in iterable]
`

### 1. Basic Syntax  
Take every item from a list, apply an operation on it, and collect the results in a new list.  


```python

numbers = [1, 2, 3, 4, 5]
cubes = [n * 3 for n in numbers]  # for each n in numbers, compute n*3

print("Original numbers:", numbers)
print("Cubes:", cubes)
```

### 2. With a Condition  
Add an if at the end to keep only the items that match a condition (this is called filtering).  

[expression for item in iterable if condition]    

```python

numbers = list(range(1, 16))
multiples_of_3 = [n for n in numbers if n % 3 == 0]  # keep n only if divisible by 3

print("Numbers 1-15:", numbers)
print("Multiples of 3:", multiples_of_3)
```

### 3. With if / else  
every item stays in the list, but its value changes based on a condition.  

 [value_if_true if condition else value_if_false for item in iterable]  


```python

scores = [55, 90, 40, 68, 72, 30]
results = ["Pass" if s >= 50 else "Fail" for s in scores]

print(results)
```

### 4. Nested Loops 
A list comprehension can loop over two (or more) iterables at once, exactly like nested for loops, but in one line.  

```python

days = ["Mon", "Tue", "Wed"]
shifts = ["AM", "PM"]

schedule = [f"{day}-{shift}" for day in days for shift in shifts]
print(schedule)
```

#### Loop vs. List Comprehension  
A list comprehension always does the same job as a for loop that builds a list — it is just shorter and more readable.  

    Traditional loop:
        cubes_loop = []
        for n in range(1, 6):
            cubes_loop.append(n ** 3)

    List comprehension (same result, one line):
        cubes_comp = [n ** 3 for n in range(1, 6)]


```python
# comparing both approaches
# Traditional loop
cubes_loop = []
for n in range(1, 6):
    cubes_loop.append(n ** 3)

# List comprehension (same result, one line)
cubes_comp = [n ** 3 for n in range(1, 6)]

print("Loop result:", cubes_loop)
print("Comprehension result:", cubes_comp)
print("Are they equal?", cubes_loop == cubes_comp)
```

## 2. Object-Oriented Programming (OOP) Basics

A class is a blueprint that bundles data (attributes) and behavior (methods) together.

- Attribute → a variable that belongs to an object (its data).
- Method → a function that belongs to an object (what it can do).
- Object (instance) → one specific thing created from the class blueprint.

You will not write many classes early on, but AI/ML frameworks (Scikit-learn estimators, PyTorch models) are all built on classes, so understanding the syntax is essential.

### 1. Creating a Class  
init is a special method called the constructor  it runs automatically when a new object is created, and sets up its attributes.  
self refers to the object itself, and must be the first parameter of every method.   

```python

class Employee:
    def __init__(self, name, salary):
        self.name = name        # attribute
        self.salary = salary    # attribute

    def annual_salary(self):    # method
        return self.salary * 12

e1 = Employee("Lina", 4000)     # creating an object (instance) of the class
print("Name:", e1.name, "| Monthly salary:", e1.salary)
print("Annual salary:", e1.annual_salary())
```

### 2. A Class with Several Methods  
A class can have as many methods as needed. Each method can read and change the object's attributes through self.  


```python

class Inventory:
    def __init__(self, item, stock=0):
        self.item = item
        self.stock = stock

    def restock(self, amount):
        self.stock += amount
        print(f"Restocked {amount}. New stock: {self.stock}")

    def sell(self, amount):
        if amount > self.stock:
            print("Not enough stock!")
        else:
            self.stock -= amount
            print(f"Sold {amount}. New stock: {self.stock}")

inv = Inventory("Notebook", 50)
inv.restock(20)
inv.sell(30)
inv.sell(1000)   # not enough stock
```

#### Important : self and __init__  
    1. __init__(self, ...)
        Runs automatically when an object is created.
        Used to set the object's starting attributes.

    2. self
        Refers to the current object.
        Must be the first parameter of every method (Python passes it automatically).
        self.balance means "this specific object's balance", not a shared value.

#### Bonus : Multiple Objects from the Same Class  
Each object created from a class has its own independent copy of the attributes. Changing one object never affects another.  
 

```python

e1 = Employee("Lina", 4000)
e2 = Employee("Omar", 5000)

print("Lina annual salary:", e1.annual_salary())
print("Omar annual salary:", e2.annual_salary())
print("Are e1 and e2 the same object?", e1 is e2)
```
