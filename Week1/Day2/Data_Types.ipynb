```python
from abc import ABC, abstractmethod

icecost = 2.50
marrgin = 0.50
class number(ABC):
    @abstractmethod
    def get(self):
        pass

    @abstractmethod
    def set(self, v):
        pass

    @abstractmethod
    def increment(self, n=1):
        pass

    @abstractmethod
    def decrement(self, n=1):
        pass

    @abstractmethod
    def __rmul__(self, other):
        pass

class number(number):
    def __init__(self, value=0):
        self._value = float(value)

    def get(self):
        return self._value

    def set(self, v):
        self._value = float(v)

    def increment(self, n=1):
        self._value += n
        return self._value

    def decrement(self, n=1):
        self._value -= n
        return self._value

    def __rmul__(self, other):
        return float(other) * self._value

    def __repr__(self):
        return str(self._value)

# use an instance with value 50
number = number(50)
total= icecost * number + marrgin * number
print("The total cost of ice cream is: $",total)
str= input("Press enter to exit")
```
