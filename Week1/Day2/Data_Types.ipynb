# Data Types


In programming, data type is an important concept.
Variables can store data of different types, and different types can do different things.
Python has the following data types built-in by default, in these categories:

Text Type:	str
Numeric Types:	int, float, complex
Sequence Types:	list, tuple, range
Mapping Type:	dict
Set Types:	set, frozenset
Boolean Type:	bool
Binary Types:	bytes, bytearray, memoryview
None Type:	NoneType


## Numeric Types:
### 1. int
A whole number (positive, negative, or zero) without a decimal point, used for counting and indexing.

```python
age =22
year =2026

print("age:" + str(age))
print("year: " + str(year))

```

###  2. Float
A number with a decimal point, used for precise measurements and mathematical calculations requiring fractions.


```python
price = 9.99
distance = 1.5

print("price: " + str(price))
print("distance: " + str(distance))
```

## Text Type:

### 1. String
sequence of characters (text) enclosed in quotes can be created using single or double quotes.

```python
name = 'Roaa'
greeting = "Hello"
print("name: " + name) 
print("greeting: " + greeting) 
print(greeting + " " + name)

print("name type: " + str(type(name)))
print("greeting type: " + str(type(greeting)))

```

## Boolean Type:
### boolean(bool)
Only two possible values: True or False (the first letter must be capitalized — this is mandatory in Python).

```python
price = 9.99
number_of_items = 5
total_cost = price * number_of_items
is_total_cost_greater_than_50 = total_cost > 50
print("total cost greater than 50: " + str(is_total_cost_greater_than_50))
print("is total cost greater than 50 type: " +str(type(is_total_cost_greater_than_50)))
```

## Sequence Types:
### 1.list
An ordered, mutable collection that can hold items of different types and allows adding, removing, modifying, sort/ reorder, or have repeated values elements.

```python
fruits = ["apple", "banana", "cherry", "date"]
print("fruits: " + str(fruits))
print(fruits[0])  # Accessing the 1st element
print(fruits[1])  # Accessing the 2nd element
print(fruits[2])  # Accessing the 3rd element
print(fruits[3])  # Accessing the 4th element

print(fruits[-1])  # Accessing the last element

#slicing
print("first three fruits: " + str(fruits[0:3]))
print (fruits[2:])
print(fruits[:])
```

#### List Methods:
1. .append(s)
it takes element s and adds it to the end of the list
2. .insert(i,s)
takes element s and adds it in index [i]
3. .remove(s)
removes the first element that match s and give error if not found in the list
4. pop(i)
remove and returns the element at index i from the list
5. .sort()
sorts the list in ascending order (permanently)
6. .reverse()
reverse the list order (permanently)
7. .index(x)
returns the the position of element x
8. .count(x)
returnss the number of occurance of element x


```python
fruits = ["apple", "banana", "cherry", "date"]
print ("fruits: " + str(fruits))
fruits.append("grapes")
fruits.append("kiwi")
print("fruits: " + str(fruits))
fruits.insert(1, "orange")  # Insert "orange" at index 1
print("fruits: " + str(fruits))
fruits.remove("banana")  # Remove "banana" from the list
print("fruits: " + str(fruits))
fruits.pop(2)  # Remove the element at index 2
print("fruits: " + str(fruits))
fruits.sort()  # Sort the list in ascending order
print("fruits: " + str(fruits))
fruits.reverse()  # Reverse the order of the list
print("fruits: " + str(fruits)) 
fruits.index("date")  # Get the index of "date"
print("index of date: " + str(fruits.index("date")))
fruits.count("apple")  # Count the occurrences of "apple"
print("count of apple: " + str(fruits.count("apple")))


```

#### looping through a list
You can visit every item in a list using a for loop.

```python
fruits = ["apple", "banana", "cherry", "date"]
for fruit in fruits:
    print(fruit)
```

#### Tuple methods
1. .count(x) — returns how many times x appears.
2. .index(x) — returns the position of the first occurrence of x.

```python
months = ("January", "February", "March", "April", "May")
print(months.count("March"))  # Count the occurrences of "March"
print(months.index("April"))  # Get the index of "April"
```

## Mapping Type:
### 1. Dictionary (dict)
Data stored as key : value pairs, surrounded by { }. Instead of accessing an item by its position (index), you access it by its name (key).

```python
person = {"name": "Roaa",
"age": 27, "city": "tubas"}
print(person)
print(type(person))

#to acces s the values in the dictionary we can use the keys
print(person["name"])
print(person["age"])
print(person["city"])
# change the value of a key
person["age"] = 26 
print(person["age"])
# add a new key-value pair to the dictionary
person["country"] = "Palestine" 
person["gender"] = "female"
print(person)

# remove a key-value pair from the dictionary
del person["city"]
print(person)
print(person.pop("gender"))
print(person)
```

#### Dict Methods:
1. .keys() — returns all keys
2. .values() — returns all values
3. .items() — returns all (key, value) pairs

```python
person = {"name": "Roaa",
"age": 27, "city": "tubas"}
print(person.keys())    
print(person.values())  
print(person.items())
```

## Set Type :
1. set
A collection of values with no specific order and no repetition, surrounded by { } (note: no keys — this is what makes it different from a dict).

Its most important feature is that it automatically removes duplicates.
Because a set is not ordered, it has no index — you can't access an item like list[0] the way you would with a list.

```python
numbers = {1, 2, 3, 2, 4, 5}
print(numbers)
#convert a list to a set to remove duplicates
num = [3, 4, 5, 6, 7]
unique_numbers = set(num)
print(unique_numbers)
# adding an item to a set
unique_numbers.add(8)
print(unique_numbers)
unique_numbers.remove(4)  # Remove an item from the set
print(unique_numbers)
unique_numbers.discard(3)  # Remove an item from the set (no error if not found)
unique_numbers.discard(10) 
print(unique_numbers)
```

## Set Operations
* '|' Union — all values from both sets
* '&' Intersection — only the values common to both sets
*  '-' Difference — values that are in the first set but not in the second

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)   # Union        -> {1, 2, 3, 4, 5, 6}
print(a & b)   # Intersection -> {3, 4}
print(a - b)   # Difference   -> {1, 2}
```
