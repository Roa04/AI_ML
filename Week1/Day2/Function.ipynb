# Python Functions
### 1. What Is a Function?
A function is a reusable block of code that performs a specific task. Instead of writing the same code repeatedly, you define it once and call it whenever needed.

Why use functions?

Avoid code duplication (DRY: Don't Repeat Yourself)

Make code easier to read and maintain

Break complex problems into smaller, manageable pieces
### 2. Basic Syntax
Every function starts with the keyword def, followed by a name, parentheses (), and a colon :. The code inside the function must be indented.

Structure:

    def function_name(parameters):

    # code to run
    
    return result

```python
def welcome_message(name):
    print("Welcome, " + name + "!")
welcome_message("roaa")
welcome_message("ahmad")
```

### 3. Parameters vs. Arguments
Parameter: Variable name in function definition
Argument: Actual value passed when calling the function

```python
def calculate_order_total(product, quantity):
    prices = {"laptop": 1200, "phone": 800, "tablet": 400}
    return prices.get(product, 0) * quantity

# Arguments: "laptop", 2
total = calculate_order_total("laptop", 2)
print("total", total)  # Output: total 2400
```

### 4. Default Parameter Values
Default values make parameters optional. If not provided, the default is used.

```python
def calculate_total(price, tax_rate=0.17):
    return price + (price * tax_rate)

print(calculate_total(80))     
print(calculate_total(130, 0.15))  
```

### 5. The return Statement
return sends a value back to the caller and immediately exits the function.

```python
# with return statement
def calculate_cube(number):
    return number ** 3
val=calculate_cube(3)
print (val)
# wihtout return statement
def message(msg):
    print(msg)

output = message("Hello, World!")
print("What message returned:", output)
```

### 6. Docstrings — Documenting Functions
Docstrings explain what a function does, its parameters, and return values.

```python
### 7. Lab Practice — Summary Statistics
def summarize_numbers(numbers):
    """Return the mean, min, and max of a list of numbers."""
    return {
        "mean": sum(numbers) / len(numbers),
        "min": min(numbers),
        "max": max(numbers)
    }

print(summarize_numbers([10, 20, 30, 40, 50]))
help(summarize_numbers)
```

### 7. Multiple Return Values
Functions can return multiple values as a tuple, dictionary, or list.

```python
def analyze_sales(sales):
    """ Analyze a list of sales figures and return summary statistics.
    Args:
        sales (list): A list of sales figures (floats or ints).
    Returns:
        dict: A dictionary containing total sales, average sales, count of sales, and highest sale.
    """
    if not sales:
        raise ValueError("Sales list cannot be empty.")
    return {
        "total": sum(sales),
        "avg": sum(sales)/len(sales),
        "count": len(sales),
        "highest": max(sales)
    }

sales = [1500, 2300, 1800, 2900, 2100]
result = analyze_sales(sales)
print(result)
```

### 8. Keyword Arguments
Pass arguments by name for clarity and flexibility in order.

```python
def create_employee(name, position, dept, salary=5000):
    return f"{ name}, {position}, {dept}, {salary}"

print(create_employee(salary=7000, name="Sara", position="Manager", dept="HR"))
print(create_employee(name="John", position="Developer", dept="IT"))
```

### 9. Mini Project — Complete Example
A professional-grade function combining everything learned.

```python
def process_grades(grades, passing=65):
    """Analyze grades and return statistics"""
    if not grades:
        return {"error": "No grades provided"}
    
    avg = sum(grades) / len(grades)
    passed = [g for g in grades if g >= passing]
    
    return {
        "total": len(grades),
        "avg": round(avg, 2),
        "highest": max(grades),
        "lowest": min(grades),
        "passed": len(passed),
        
    }

scores = [58, 72, 91, 45, 88, 76, 62, 95, 83, 67, 49, 77]
results = process_grades(scores)
print (results)
print(f"Total: {results['total']}")
print(f"Average: {results['avg']}")
print(f"Passed: {results['passed']}")
```
