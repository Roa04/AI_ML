## 1. Visualization
Charts make patterns visible that tables can hide — like trends, clusters, and outliers. Visualization is both a way to explore data during analysis and to communicate findings, forming the core of Exploratory Data Analysis (EDA).

## 2. The Matplotlib Basics
main plotting library. The process is simple: import pyplot, call a plot function, add labels and a title, then show the chart. Labels are essential — without them, a chart is confusing instead of informative.

```python
x= range(2000,2010)
y= [0.8,0.9,0.921,0.987,0.56,.89,.19,.47,.23,.62]
plt.plot(x,y)
# name the xand y axsis
plt.xlabel("year")
plt.ylabel("performance")
```

```python
year = range(2000,2010)
performance = [0.8,0.9,0.921,0.987,0.56,.89,.19,.47,.23,.62]
success      =[0.89,0.99,1,1.3,0.63,.93,.2,.78,.33,.72]

print(type(year))
print(len(year))
print(list(year))

print(type(performance))
print(len(performance))
print(performance)
```

```python
# plot two lines (performance and success) on the same chart
plt.plot(year,performance)
plt.plot(year,success)
plt.xlabel("year")
plt.ylabel("performance & success")

plt.title ("Performance vs Success")
```

```python
# same plot as above, but with a legend to label each line
year = range(2000,2010)
performance = [0.8,0.9,0.921,0.987,0.56,.89,.19,.47,.23,.62]
success      =[0.89,0.99,1,1.3,0.63,.93,.2,.78,.33,.72]

plt.plot(year,performance)
plt.plot(year,success)
plt.xlabel("year")
plt.ylabel("performance & success")

plt.title ("Performance vs Success")
plt.legend(["performance","success"])
```

```python
# to add edges

plt.plot(year,performance, marker='*')
plt.plot(year,success, marker='o')
plt.xlabel("year")
plt.ylabel("performance & success")

plt.title ("Performance vs Success")
plt.legend(["performance","success"])
```

```python
# set custom colors for each line

plt.plot(year,performance, marker='*',c='b')
plt.plot(year,success, marker='o',c='r')
plt.xlabel("year")
plt.ylabel("performance & success")

plt.title ("Performance vs Success")
plt.legend(["performance","success"])
```

```python
# change the style of lines

plt.plot(year,performance, marker='*',c='b',ls='--')
plt.plot(year,success, marker='o',c='r')
plt.xlabel("year")
plt.ylabel("performance & success")

plt.title ("Performance vs Success")
plt.legend(["performance","success"])
```

```python
# change the size of the edge

plt.plot(year,performance, marker='*',c='b',ls='--',ms='7')
plt.plot(year,success, marker='o',c='r',ms='4')
plt.xlabel("year")
plt.ylabel("performance & success")

plt.title ("Performance vs Success")
plt.legend(["performance","success"])
```

```python

# control marker edge width (mew), edge color (mec), and line transparency (alpha)


plt.plot(year,performance, marker='*',c='b',ls='--',ms='7',mew='2',mec='purple')
plt.plot(year,success, marker='o',c='r',ms='4',alpha=.4)
plt.xlabel("year")
plt.ylabel("performance & success")

plt.title ("Performance vs Success")
plt.legend(["performance","success"])
```

```python
# shorthand string format for color+marker+linestyle
plt.plot(year,performance,'b-x')
plt.plot(year,success,'r-o')
plt.xlabel("year")
plt.ylabel("performance & success")

plt.title ("Performance vs Success")
plt.legend(["performance","success"])
```
