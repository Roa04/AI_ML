## 5.1  Bivariate Analysis
Bivariate analysis examines the relationship between two variables — the step where genuine insight usually 
appears, because ML is fundamentally about relationships between features and a target. A scatter plot reveals the 
relationship between two numeric variables; grouped box plots reveal how a numeric variable differs across 
categories.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("healthcare-dataset-stroke-data.csv")
df
```

```python
df.isnull().sum()
```

```python

df['bmi'] = df['bmi'].fillna(df['bmi'].mean())
df.isnull().sum()
```

```python
sns.scatterplot(data=df, x="age", y="bmi")
```

```python
sns.boxplot(data=df, x="gender", y="avg_glucose_level")
```

## 5.2  Correlation
Correlation measures how strongly two numeric variables move together, on a scale from -1 to +1: +1 is a perfect 
positive relationship, -1 a perfect negative one, and 0 no linear relationship. A correlation heatmap shows every 
pair of variables' correlations at once — one of the most informative single charts in all of EDA.

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
```

<p style="color: skyblue;">Correlation is not causation: two variables moving together does not mean one causes the other. EDA identifies 
relationships; it never proves cause.</p>

## 5.3  The Pairplot
sns.pairplot() plots every numeric variable against every other in a single grid, with distributions on the diagonal — 
a fast way to scan an entire dataset's relationships at once, useful early in EDA to decide where to look closer.

```python
sns.pairplot(df,hue="age")
```

```python
sns.pairplot(df,hue="stroke")
```

## 5.4  Data Storytelling
EDA produces charts, but the deliverable is understanding, communicated clearly. Data storytelling means turning 
the analysis into a narrative: what the data contains, what patterns and problems were found, and what they imply 
for modeling. This is the same communication skill emphasized throughout the program — a finding nobody 
understands is not a finding.
