## 1.Represent three data samples as a (3 × features) NumPy matrix.

```python
import numpy as np
matrix= np.array([[50,62,49],[38,84,61],[21,98,75]])
```

## 2. Compute the dot product of one sample vector with a weight vector by hand, then verify it with (np.dot.)

```python
weight = np.array([2.0, 0.5, 1.0])  
sample = matrix[0]

byhand = (sample[0]*weight[0]) + (sample[1]*weight[1]) + (sample[2]*weight[2])
print("Computed by hand:", byhand)

dotv= np.dot(sample,weight)
print("prediction:",dotv)
```

## 3. Use matrix multiplication (@) to produce a prediction for all three samples at once.

```python
w= np.array([-2,5,8])
matrix= np.array([[50,62,49],[38,84,61],[21,98,75]])

print("matrix multiplication: ",matrix @ w)
```

## 4. Deliberately create a shape-mismatch error, read the message, and explain in Markdown why it occurred and how to fix it.

```python
w_bad = np.array([-2, 5, 8, 4,7,3]) 
matrix= np.array([[50,62,49],[38,84,61],[21,98,75]])


matrix @ w_bad
```

### Why it occurred: 
 matrix has 3 columns (features), and w_bad here has 6 elements, not 3. Matrix multiplication requires the number of columns in the left operand to equal the number of elements in the right operand (the vector). 6 ≠ 3, so it fails.

### How to fix it:
 trim w_bad down to 3 elements, e.g. np.array([-2, 5, 8]), or make the matrix  6 columns (6 features per sample) instead of 3.

```python
w1 = np.array([-2, 5, 8, 4,7,3]) 
matrix1 = np.array([
    [50, 62, 49, 30, 15, 20],
    [38, 84, 61, 45, 22, 10],
    [21, 98, 75, 60, 33, 5]
])
print("the correct way1: \n",matrix1 @ w1)


# or

w2=np.array([7,4,3])
matrix2= np.array([[50,62,49],[38,84,61],[21,98,75]])
print("the correct way2: \n",matrix2 @ w2)


```
