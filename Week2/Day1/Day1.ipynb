* Why descriptive statistics precedes modeling
* Central tendency: mean, median, mode
* Spread: range, variance, standard deviation, IQR
* Percentiles and quartiles

### 1.1 Statistic befor modeling
A machine learning model is basically a summary of patterns in data. To judge if a model makes sense, you first need to understand the dataset — its average, variation, and overall shape. Descriptive statistics give you the precise language to do that, forming the foundation for exploring and evaluating data.

### 1.2  Measures of Central Tendency

* Mean: The arithmetic average ,Symmetric data with no extreme outliers
* Median: The middle value when sorted, Skewed data or data with outliers (more robust)
* Mode: The most frequent value, Categorical data, or finding the most common value

```python
import numpy as np
data =np.array([5,7,4,2,6,9,5])
print("mean: ",np.mean(data))
print("median: ",np.median(data))
```

```python
import scipy.stats as stats

print("mode: ",stats.mode(data))
```

### 1.3  Measures of Spread
Spread shows how far data points lie from the average, Two datasets can have the same mean but look very different if one is tightly clustered and the other widely scattered.


* Range: Maximum minus minimum — simple but very sensitive to outliers
* Variance: The average squared distance from the mean
* Standard deviation: The square root of variance — spread in the same units as the data
* IQR (interquartile range): The range of the middle 50% of the data (Q3 - Q1), robust to outliers

```python



data2= np.array([4,8,9,3,5,6,7])
print("max: ",data.max())
print("min: ",data.min())
print("range: ", data.max()-data.min())
print("Standard deviation: ", np.std(data))
print("variance: ", np.var(data))
q1, q3 = np.percentile(data, [25, 75])
iqr = q3 - q1
print("IQR:", iqr)

```

* Step 2: Compute its mean, median, mode, standard deviation, and IQR.

```python
mean= np.mean(df["fare"])
median=np.median(df["fare"])
mode =stats.mode(df["fare"])
stde=np.std(df["fare"])
q1, q3 = np.percentile(df["fare"], [25, 75])
iqr = q3 - q1
print("mean: ",mean)
print("median: ",median)
print("mode: ",mode)
print("standard deviation: ",stde)
print("IQR:", iqr)
```

* Step 3: Identify whether the mean or median better represents a 'typical' value, and justify the choice.
